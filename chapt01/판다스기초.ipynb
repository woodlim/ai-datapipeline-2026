#### Series

- 1차원 데이터를 표현하는  자료구조. `인덱스`가 붙어있는 한줄짜리 데이터. 1차원 배열과 흡사

````python
np_scores = np.array([85, 90, 95, 80])

np_scores
````

````python
scores = pd.Series([85, 90, 95, 80])

scores
````

````python
scores[0]
````

- 시리즈 작성 시 인덱스를 커스터마이징 할 수 있음

````python
scores = pd.Series(
    [85, 90, 95, 80],
    index = ['홍길동', '김철수', '이영희', '박민수']
)

scores
````

````python
# 기존 인덱스는 사용 불가
scores[0]
````

````python
# 인덱스가 숫자에서 문자열로 변경.
scores['홍길동']
````

#### DataFrame

- Pandas 2차원 표형태 데이터 표현하는 자료구조
- Pandas의 기본 자료구조형
- 엑셀, CSV(엑셀형태 텍스트파일), DB 테이블과 거의 동일한 구조

````python
# DataFrame을 직접 생성할때 사용.
#파이썬 딕셔너리 사용
raw_data = {
    'name' : ['홍길동', '김철수', '이영희', '박민수','성유고','애슐리'],
    'age' : [20, 21, 22, 23, 30, 29],
    'major' : ['데이터분석', '웹개발', '인공지능', '임베디드', '인공지능', '웹개발'],
    'score' : [85, 90, 95, 80, 90, 100]

}

raw_data
````

````python
# 일반적으로 DataFrame 줄여서 df로 변수명 많이 활용
df = pd.DataFrame(raw_data)

df
````
